# Research Question

How do production budget, movie genre, lead actors, language, and country of production influence a film's financial and critical success — measured by box office revenue and IMDb ratings — and does this relationship differ between US and non-US productions, and across the pre- and post-streaming eras (before and after 2015)?

Specifically, we examine whether the budget-success relationship has weakened as streaming platforms have disrupted theatrical revenue models, and whether Hollywood's historically dominant global distribution network produces a stronger budget-revenue association compared to non-US film industries. We are also interested in whether lower-budget international films in certain genres can achieve comparable critical success (IMDb ratings) despite lower revenue, suggesting that financial and critical success are not always aligned.

The key variables include production budget (numerical), genre (categorical), country of production (categorical), lead actor (categorical), original language (categorical), release year (numerical, used to define pre/post-2015 periods), box office revenue (numerical), and IMDb rating (numerical). We will use correlation analysis, group comparisons, and interaction-based statistical models to examine these relationships. This is an inferential data analysis task that builds on and extends prior work by incorporating geographic, temporal, and talent-related dimensions that the existing literature has largely overlooked.

## Background and Prior Work

The film industry is strongly influenced by production budgets, which affect everything from casting and visual effects to marketing and distribution. Different movie genres tend to require different levels of funding. Research on box office revenue prediction has confirmed that production budget is the single strongest predictor of a film's financial performance, with action and adventure genres consistently commanding the highest budgets while drama and horror operate at far lower cost levels.<sup>1</sup>

Previous peer-reviewed work has also shown that genre plays a major role in determining production cost and financial outcomes. Shahid and Islam (2023) demonstrated that a genre's popularity at the time of release is a strong predictor of box office returns, and that drama films consistently carry lower budgets than action or adventure films — underscoring how budget and genre interact to shape financial outcomes.<sup>2</sup>

When it comes to critical success, the picture is different. A study analyzing IMDb user voting data found no significant correlation between production budget and average user ratings, suggesting that audience satisfaction is driven by factors such as storytelling quality and emotional resonance rather than spending alone.<sup>3</sup> This supports our hypothesis that financial and critical success are not always aligned, and that lower-budget international films may achieve high ratings despite modest revenues.

Despite this body of work, several important gaps remain. First, most existing studies treat the global film industry as a monolith and focus almost exclusively on Hollywood. However, industry data show that international revenues for US studio films now account for over 70% of Hollywood's total box office take, and that non-US industries operate under fundamentally different distribution conditions.<sup>4</sup> Cross-national comparisons of how budget, language, and talent translate to success are therefore critical but largely absent from the literature.

Second, few studies account for the structural disruption caused by streaming platforms. Research published in Applied Economics found that the launch of Netflix streaming in the US was associated with a 14–17% reduction in theatrical box office revenues, establishing a clear empirical link between streaming availability and declining theatrical performance.<sup>5</sup> This suggests that the budget-revenue relationship documented in pre-streaming studies may no longer hold post-2015, yet most prior analyses ignore this temporal shift. Our project directly addresses both gaps by incorporating country of production and original language as comparative variables, and by splitting our analysis across pre- and post-2015 time periods.

# Hypothesis

We hypothesize that movies with larger production budgets, globally recognized actors, and English-language releases will generally earn higher box office revenue due to wider international distribution and marketing reach. Specifically, we expect this budget-revenue relationship to be stronger for US films than for non-US films, because Hollywood benefits from an established global distribution and marketing infrastructure that amplifies returns on high-budget productions — a network advantage that non-US industries generally lack.

We further hypothesize that this budget-revenue association has weakened in the post-2015 period compared to the pre-2015 era, as the proliferation of streaming platforms has reduced audiences' dependence on theatrical releases, particularly for lower-budget films that would previously have relied on box office performance as their primary revenue source.

In contrast, we expect IMDb ratings to remain largely independent of production budget regardless of country of production or time period, consistent with prior empirical findings showing no significant correlation between budget and average user ratings. We also predict that some lower-budget international films, particularly in drama or thriller genres, may receive disproportionately high IMDb ratings relative to their revenue — suggesting that critical success and financial success are not always strongly aligned.

Among genres, we expect action, adventure, and fantasy films to show the strongest positive budget-revenue correlation, since these genres depend most heavily on spectacle and wide theatrical distribution to justify their investment, while drama and horror — genres that can succeed critically and commercially on low budgets — will show weaker or more variable correlations.

### Dataset #1 — Movie Industry (Daniel Grijalvas)

This dataset contains 6,820 films released between 1986 and 2016 with 15 variables covering production details, financial performance, and audience reception. The core financial variables are `budget` (production cost in USD) and `gross` (total theatrical box office revenue in USD). The `score` variable is an IMDb user rating on a 0–10 scale and serves as our primary measure of critical/audience success.

Budget values range widely — from a few thousand dollars for micro-budget films to over $300 million for major blockbusters — with a heavily right-skewed distribution. We address this by log-transforming both budget and gross after adjusting for inflation using BLS CPI-U data (base year: 2020), since raw dollar values are not comparable across four decades. A significant number of entries have missing or zero budget values (likely unreported rather than truly zero-budget). We drop these rows and document the extent of missingness below. `votes` are excluded from predictive modeling since they are a post-release variable.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# --- Load raw data ---
df = pd.read_csv('data/00-raw/movies.csv')
print('Shape:', df.shape)
df.head()
```

```python
# --- Visualize missingness ---
missing_values = df.isnull().sum()
missing_percent = (df.isnull().mean() * 100).sort_values(ascending=False)
missing_table = pd.DataFrame({
    "Missing Values": missing_values,
    "Percent Missing": missing_percent
})
missing_table.sort_values(by="Percent Missing", ascending=False)
```

```python
# --- Parse release date, drop post-release variable ---
df['release_date'] = pd.to_datetime(
    df['released'].str.extract(r'^([^(]+)').squeeze().str.strip(),
    errors='coerce'
)
df = df.drop(columns=['released', 'votes'])

# --- Strip whitespace from string columns ---
str_cols = df.select_dtypes(include='object').columns
df[str_cols] = df[str_cols].apply(lambda s: s.str.strip())

# --- Cast numeric columns ---
for col in ['budget', 'gross', 'runtime']:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# --- Flag zero budgets/gross as missing ---
df.loc[df['budget'] == 0, 'budget'] = np.nan
df.loc[df['gross'] == 0, 'gross'] = np.nan

print('After cleaning, shape:', df.shape)
print('Missing budget:', df['budget'].isnull().sum())
print('Missing gross:', df['gross'].isnull().sum())

# --- Drop rows missing budget (our primary predictor) ---
df = df.dropna(subset=['budget'])
print('After dropping missing budget, shape:', df.shape)
```

```python
# --- CPI-U inflation adjustment to 2020 dollars ---
cpi = {
    1980: 82.4, 1981: 90.9, 1982: 96.5, 1983: 99.6, 1984: 103.9,
    1985: 107.6, 1986: 109.6, 1987: 113.6, 1988: 118.3, 1989: 124.0,
    1990: 130.7, 1991: 136.2, 1992: 140.3, 1993: 144.5, 1994: 148.2,
    1995: 152.4, 1996: 156.9, 1997: 160.5, 1998: 163.0, 1999: 166.6,
    2000: 172.2, 2001: 177.1, 2002: 179.9, 2003: 184.0, 2004: 188.9,
    2005: 195.3, 2006: 201.6, 2007: 207.3, 2008: 215.3, 2009: 214.5,
    2010: 218.1, 2011: 224.9, 2012: 229.6, 2013: 233.0, 2014: 236.7,
    2015: 237.0, 2016: 240.0, 2017: 245.1, 2018: 251.1, 2019: 255.7,
    2020: 258.8
}

BASE_CPI = cpi[2020]
df['cpi'] = df['year'].map(cpi)
df['budget_real'] = df['budget'] * (BASE_CPI / df['cpi'])
df['gross_real'] = df['gross'] * (BASE_CPI / df['cpi'])
df = df.drop(columns=['cpi'])

# --- Log-transform to handle right skew ---
df['log_budget'] = np.log1p(df['budget_real'])
df['log_gross'] = np.log1p(df['gross_real'])

# --- Create era and US indicator columns ---
df['era'] = df['year'].apply(lambda y: 'post_2015' if y >= 2015 else 'pre_2015')
df['us_film'] = df['country'].apply(lambda c: 'US' if c == 'United States' else 'non-US')

print('Final Dataset 1 shape:', df.shape)
df.describe()
```

```python
# --- Visualize distributions to check for outliers ---
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

df['log_budget'].hist(bins=40, ax=axes[0], color='steelblue', edgecolor='white')
axes[0].set_title('Log Budget (2020 USD)')
axes[0].set_xlabel('log(budget)')

df['log_gross'].hist(bins=40, ax=axes[1], color='coral', edgecolor='white')
axes[1].set_title('Log Gross Revenue (2020 USD)')
axes[1].set_xlabel('log(gross)')

df['score'].hist(bins=30, ax=axes[2], color='mediumseagreen', edgecolor='white')
axes[2].set_title('IMDb Score Distribution')
axes[2].set_xlabel('IMDb Score (0-10)')

plt.tight_layout()
plt.show()
```

```python
# --- Basic EDA: Summary Statistics ---
print("\nDataset Shape:")
print(df.shape)
print("\nSummary Statistics:")
print(df[['budget_real', 'gross_real', 'runtime', 'score', 'log_budget', 'log_gross']].describe())

# Categorical summaries
print("\nMovies by Era")
print(df['era'].value_counts())
print("\nUS vs Non-US Films")
print(df['us_film'].value_counts())
```

```python
# --- Distribution Plots ---
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

axes[0,0].hist(df['budget_real'].dropna(), bins=40, edgecolor='white')
axes[0,0].set_title('Budget Distribution (2020 USD)')
axes[0,0].set_xlabel('Budget')
axes[0,0].set_ylabel('Frequency')

axes[0,1].hist(df['gross_real'].dropna(), bins=40, edgecolor='white')
axes[0,1].set_title('Gross Revenue Distribution (2020 USD)')
axes[0,1].set_xlabel('Gross Revenue')
axes[0,1].set_ylabel('Frequency')

axes[1,0].hist(df['score'].dropna(), bins=30, edgecolor='white')
axes[1,0].set_title('IMDb Score Distribution')
axes[1,0].set_xlabel('IMDb Score')
axes[1,0].set_ylabel('Frequency')

axes[1,1].hist(df['runtime'].dropna(), bins=30, edgecolor='white')
axes[1,1].set_title('Runtime Distribution')
axes[1,1].set_xlabel('Runtime (minutes)')
axes[1,1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()

# --- Save processed Dataset 1 ---
df.reset_index(drop=True).to_csv('data/02-processed/movies_clean.csv', index=False)
print('Dataset 1 saved to data/02-processed/movies_clean.csv')
```

### Merging the Datasets (Ketan)

We merge on movie title and release year after standardizing column names. An inner join is used to produce a combined dataset with full information from both sources. Unmatched rows are noted for reference.

```python
# Loading both datasets
df1 = pd.read_csv('data/00-raw/movies.csv')
df2 = pd.read_csv('data/00-raw/movies_metadata.csv')

print(f"Dataset 1 (movies.csv): {df1.shape}")
print(f"Dataset 2 (movies_metadata.csv): {df2.shape}")
```

```python
# Preparing merge keys
df1['merge_title'] = df1['name'].str.lower().str.strip()
df1['merge_year'] = df1['year']

df2['merge_title'] = df2['title'].str.lower().str.strip()
df2['merge_year'] = pd.to_datetime(df2['release_date'], errors='coerce').dt.year

# Dropping rows where merge keys are missing
df1_clean = df1.dropna(subset=['merge_title', 'merge_year']).copy()
df2_clean = df2.dropna(subset=['merge_title', 'merge_year']).copy()
df2_clean['merge_year'] = df2_clean['merge_year'].astype(int)

# Picking relevant columns from each before merging
df1_cols = ['merge_title', 'merge_year', 'name', 'year', 'genre', 'rating',
            'score', 'votes', 'director', 'star', 'country', 'budget',
            'gross', 'company', 'runtime']
df2_cols = ['merge_title', 'merge_year', 'genres', 'original_language',
            'production_countries', 'popularity', 'revenue', 'vote_average',
            'vote_count', 'cast']

# Inner join on title + year
merged = pd.merge(
    df1_clean[df1_cols],
    df2_clean[df2_cols],
    on=['merge_title', 'merge_year'],
    how='inner',
    suffixes=('_ds1', '_ds2')
)

# Dropping the merge helper cols
merged = merged.drop(columns=['merge_title', 'merge_year'])

print(f"\nMerged dataset: {merged.shape}")
print(f"Lost from DS1: {len(df1) - len(merged)} rows")
print(f"Lost from DS2: {len(df2) - len(merged)} rows")
print(f"\nColumns:\n{merged.columns.tolist()}")
merged.head()
```

```python
# Add derived columns needed for EDA
merged['log_budget'] = np.log1p(merged['budget'])
merged['log_gross'] = np.log1p(merged['gross'])
merged['era'] = merged['year'].apply(lambda y: 'Post-2015' if y >= 2015 else 'Pre-2015')
merged['is_us'] = merged['country'] == 'United States'

# Save merged dataset
merged.to_csv('data/02-processed/merged_movies.csv', index=False)
print('Merged dataset saved to data/02-processed/merged_movies.csv')
```

#### Section 3: Budget vs. Gross Revenue by Genre

A central question of our project is whether budget predicts revenue. We plot budget vs. gross on a log-log scale to see the relationship across genres. We restrict to the top 6 genres for visual clarity since smaller genres have too few observations to interpret meaningfully. The dashed line represents the break-even point (gross = budget). Most films above the line are profitable on a raw revenue basis. The overall log-log correlation of r ≈ 0.67 confirms a strong positive relationship between budget and gross, consistent with prior literature.

```python
top_genres = df['genre'].value_counts().head(6).index.tolist()
df_plot = df[df['genre'].isin(top_genres)].dropna(subset=['budget', 'gross'])

fig, ax = plt.subplots(figsize=(10, 6))
sns.scatterplot(data=df_plot, x='budget', y='gross', hue='genre',
                alpha=0.6, s=30, palette='tab10', ax=ax)
ax.set_xscale('log')
ax.set_yscale('log')
ax.set_xlabel('Production Budget (USD, log scale)')
ax.set_ylabel('Gross Revenue (USD, log scale)')
ax.set_title('Budget vs. Gross Revenue by Genre (Top 6 Genres)')

# Reference line: gross = budget (break-even, approximately)
xlim = ax.get_xlim()
ax.plot(xlim, xlim, 'k--', alpha=0.4, label='Gross = Budget')
ax.legend(bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()

# Quick correlation in log space
log_corr = np.log10(df_plot['budget']).corr(np.log10(df_plot['gross']))
print(f"Log-log correlation between budget and gross: r = {log_corr:.3f}")
```

We compute the pairwise Pearson correlations between budget, gross revenue, and IMDb score. As expected, budget and gross are strongly correlated (r ≈ 0.74), confirming the budget-revenue relationship. Importantly, budget shows essentially no correlation with IMDb score (r ≈ 0.07), consistent with prior empirical findings and our hypothesis that critical and financial success are driven by different factors.

#### Section 5: US vs. Non-US Budget-Revenue Relationship (Jackie)

We compare the budget-revenue relationship separately for US and non-US productions. Both groups show a positive log-linear relationship, but US films (r ≈ 0.67) have a modestly stronger correlation than non-US films (r ≈ 0.59), providing preliminary support for our hypothesis that Hollywood's distribution infrastructure amplifies returns on high-budget investments. The overlap in regression lines suggests the difference is real but not dramatic, consistent with the global reach of major non-US industries such as Bollywood and Korean cinema.

```python
g = sns.lmplot(
    data=df,
    x='log_budget',
    y='log_gross',
    hue='is_us',
    scatter_kws={'alpha': 0.4},
    height=6,
    aspect=1.2
)
plt.xlabel('Log Budget')
plt.ylabel('Log Gross Revenue')
plt.title('Budget-Revenue Relationship: US vs Non-US')
plt.show()

# Correlations
us_corr = df[df['is_us']]['log_budget'].corr(df[df['is_us']]['log_gross'])
nonus_corr = df[~df['is_us']]['log_budget'].corr(df[~df['is_us']]['log_gross'])
print(f"US films correlation: {us_corr:.3f}")
print(f"Non-US films correlation: {nonus_corr:.3f}")
```

#### Section 6: Pre/Post-2015 (Streaming Era) Comparison (Jackie)

We split the data at 2015 to test whether the budget-revenue relationship has weakened in the streaming era. Contrary to our hypothesis, the post-2015 correlation (r ≈ 0.71) is slightly higher than the pre-2015 correlation (r ≈ 0.64). This may reflect survivor bias in the post-2015 theatrical market — only high-budget, wide-release films consistently appear in this dataset's theatrical gross figures after streaming became dominant — or it may indicate that the strongest budget-revenue relationship now belongs exclusively to blockbuster-scale productions.

```python
g = sns.lmplot(
    data=df,
    x='log_budget',
    y='log_gross',
    hue='era',
    scatter_kws={'alpha': 0.4},
    height=6,
    aspect=1.2
)
plt.xlabel('Log Budget')
plt.ylabel('Log Gross Revenue')
plt.title('Budget-Revenue Relationship by Era (Pre vs. Post-2015)')
plt.show()

# Correlations by era
pre_corr = df[df['era'] == 'Pre-2015']['log_budget'].corr(
    df[df['era'] == 'Pre-2015']['log_gross']
)
post_corr = df[df['era'] == 'Post-2015']['log_budget'].corr(
    df[df['era'] == 'Post-2015']['log_gross']
)
print(f"Pre-2015 correlation: {pre_corr:.3f}")
print(f"Post-2015 correlation: {post_corr:.3f}")
```

#### Section 7: Budget vs. IMDb Score (Jackie)

We directly test our hypothesis that IMDb score is independent of production budget. The scatter plot and near-zero regression slope (r ≈ 0.00) confirm that budget has essentially no linear relationship with audience ratings. This is consistent with prior empirical findings and supports the idea that critical and financial success are driven by fundamentally different factors — storytelling quality and emotional resonance versus marketing spend and distribution reach.

```python
fig, ax = plt.subplots(figsize=(8, 6))
sns.scatterplot(data=df, x='log_budget', y='score', alpha=0.5, ax=ax)
sns.regplot(data=df, x='log_budget', y='score', scatter=False, color='red', ax=ax)
ax.set_xlabel('Log Budget')
ax.set_ylabel('IMDb Score')
ax.set_title('Budget vs. IMDb Rating')
plt.show()

rating_corr = df['log_budget'].corr(df['score'])
print(f"Budget-IMDb correlation: {rating_corr:.3f}")
```

#### Section 8: Genre-Level Budget-Revenue Correlations

We compute the budget-gross correlation separately for each genre to test our hypothesis that action, adventure, and fantasy will show the strongest associations. The results largely confirm this: Action (r ≈ 0.76), Animation (r ≈ 0.72), and Fantasy (r ≈ 0.70) show among the highest correlations, consistent with their dependence on spectacle and wide theatrical distribution. Horror (r ≈ 0.37) and Drama (r ≈ 0.53) show weaker correlations, consistent with their ability to succeed commercially on low budgets.

```python
genre_corrs = (
    df.groupby('genre')
    .apply(lambda x: x['log_budget'].corr(x['log_gross']))
    .sort_values(ascending=False)
)
print(genre_corrs)
```

#### Section 9: OLS Regression Model (Jackie)

We fit an OLS regression model predicting log gross revenue from log budget, US vs. non-US indicator, era, their interactions, genre fixed effects, and IMDb score. The model explains approximately 50% of the variance in log gross revenue (R² ≈ 0.507). Log budget is the dominant predictor (coef ≈ 0.87, p < 0.001). The interaction terms for US status and era are not statistically significant at conventional levels, suggesting that while raw correlations differ, the budget-revenue slope does not significantly differ between US/non-US or pre/post-2015 films after controlling for genre and score. IMDb score has a small but significant positive effect on gross (coef ≈ 0.20, p < 0.001).

```python
import statsmodels.formula.api as smf

model = smf.ols(
    """
    log_gross ~ log_budget
    + C(is_us)
    + C(era)
    + log_budget:C(is_us)
    + log_budget:C(era)
    + C(genre)
    + score
    """,
    data=df
).fit()

print(model.summary())
```

## Ethics

### A. Data Collection

- **A.1 Informed consent:** This project uses publicly available datasets (IMDb via Kaggle), so no human subjects are directly involved and informed consent is not required.
- **A.2 Collection bias:** Both datasets are biased toward popular, Western, and high-budget films, which may underrepresent non-US industries such as Bollywood, Nollywood, or Korean cinema. This directly affects our US vs. non-US comparison. We explicitly report sample sizes by country group and acknowledge this imbalance when interpreting results.
- **A.3 Limit PII exposure:** The datasets do not include personally identifiable information. All data points represent films, not individuals, so privacy risks are minimal.
- **A.4 Downstream bias mitigation:** Because our datasets skew toward Hollywood productions, conclusions drawn from non-US subgroups may be less statistically reliable. We flag this limitation in our results and avoid overgeneralizing from underrepresented groups.

### B. Data Storage

- **B.1 Data security:** Data is stored in a private GitHub repository and on local machines accessible only to team members.
- **B.2 Right to be forgotten:** All data used is publicly available and contains no personal information, so the right to be forgotten is not directly applicable.
- **B.3 Data retention plan:** Data will only be used for the duration of this academic project. All local copies will be deleted upon course completion and the repository archived.

### C. Analysis

- **C.1 Missing perspectives:** Our datasets significantly underrepresent non-English-language and non-Western film industries. Findings from non-US subgroups will be interpreted cautiously with sample sizes reported.
- **C.2 Dataset bias:** We examine potential bias in genre representation and budget reporting, including checking whether certain genres or countries are systematically underrepresented.

- **C.4 Privacy in analysis:** No personal data will be displayed or used. Analysis operates at the film level, not the individual level.
- **C.5 Auditability:** All analysis is documented in reproducible Jupyter notebooks, version-controlled on GitHub.

### D. Modeling

- **D.1 Proxy discrimination:** Country of production and original language are used as structural variables, not as proxies for race or ethnicity. We will not frame results in ways that imply value judgments about different film industries or cultures.
- **D.2 Fairness across groups:** We report results separately for US vs. non-US films and for pre/post-2015 periods, so patterns specific to smaller groups are not masked.
- **D.3 Metric selection:** We acknowledge that box office revenue does not capture streaming income or home video sales, and that IMDb ratings may reflect demographic biases in the IMDb user base.
- **D.4 Explainability:** Methods and results will be clearly explained in plain language alongside technical details, with visualizations to aid accessibility.
- **D.5 Communicate limitations:** We explicitly state dataset limitations, potential sources of bias, and boundaries on generalizability in both our notebooks and final report.

- **E.2 Redress:** If errors are identified, we will revise our analysis, correct affected conclusions, and clearly document what changed and why.
- **E.3 Roll back:** All analysis is version-controlled on GitHub, allowing rollback to any prior version.
